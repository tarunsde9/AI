```python
# Import SentenceTransformer to generate text embeddings
from sentence_transformers import SentenceTransformer

# Import cosine_similarity to compare document and query embeddings
from sklearn.metrics.pairwise import cosine_similarity

# Import pandas to organize documents and search results
import pandas as pd

# Import numpy for numerical operations
import numpy as np

print("All libraries imported successfully!")
```

```python
# Create a collection of documents for semantic search
documents = [
    # Python Programming
    "Python is a popular programming language used for automation and web development.",
    "Python lists are used to store multiple items in a single variable.",
    "Python dictionaries store data in key-value pairs.",
    "Functions in Python help organize reusable blocks of code.",
    "Pandas is a Python library used for data analysis and manipulation.",

    # Java Development
    "Java is an object-oriented programming language used for enterprise applications.",
    "Spring Boot simplifies the development of Java backend applications.",
    "Java inheritance allows one class to reuse properties of another class.",
    "Java streams help process collections of data efficiently.",
    "Multithreading allows Java programs to execute multiple tasks concurrently.",

    # Machine Learning
    "Machine learning enables computers to learn patterns from data.",
    "Linear regression predicts continuous numerical values.",
    "Classification algorithms predict categories such as spam or not spam.",
    "Decision trees make predictions using a tree-like structure.",
    "Model evaluation measures how well a machine learning model performs.",

    # Artificial Intelligence
    "Artificial intelligence enables machines to perform tasks requiring human intelligence.",
    "Natural language processing helps computers understand human language.",
    "Chatbots use artificial intelligence to communicate with users.",
    "Neural networks are inspired by the structure of the human brain.",
    "Generative AI can create text, images, and other types of content.",

    # Databases
    "SQL is used to retrieve and manage data in relational databases.",
    "MongoDB is a NoSQL database that stores data in document format.",
    "Database indexing improves the speed of data retrieval.",
    "Database normalization reduces data duplication.",
    "Transactions ensure database operations maintain data consistency.",

    # Cloud Computing
    "AWS provides cloud services such as computing, storage, and networking.",
    "Amazon S3 is used to store files and objects in the cloud.",
    "AWS EC2 provides virtual servers for running applications.",
    "Cloud computing allows businesses to access computing resources on demand.",
    "Serverless computing allows developers to run code without managing servers.",

    # Docker and DevOps
    "Docker packages applications and their dependencies into containers.",
    "Docker images are templates used to create containers.",
    "Docker Compose manages applications containing multiple containers.",
    "Kubernetes orchestrates containers across multiple machines.",
    "Continuous integration automates building and testing software."
]

# Display the total number of documents
print("Total documents:", len(documents))

# Display the first five documents
for index, document in enumerate(documents[:5], start=1):
    print(f"{index}. {document}")
```

```python
# Create a DataFrame from our document collection
df = pd.DataFrame({
    "document_id": range(1, len(documents) + 1),
    "text": documents
})

# Display the first five documents
display(df.head())

# Display the total number of documents
print("Total documents:", len(df))
```

```python
# Load the pre-trained sentence embedding model
model = SentenceTransformer("all-MiniLM-L6-v2")

# Confirm that the model is loaded
print("Embedding model loaded successfully!")
```

```python
# Convert all document texts into numerical embeddings
document_embeddings = model.encode(
    df["text"].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

# Display the shape of the generated embeddings
print("Embedding shape:", document_embeddings.shape)

# Display the embedding of the first document
print("\nFirst document embedding:")
print(document_embeddings[0])
```

```python
# Define a sample user search query
query = "How can I build applications using Python?"

# Convert the query into a numerical embedding
query_embedding = model.encode(
    query,
    convert_to_numpy=True
)

# Display the query embedding shape
print("Query embedding shape:", query_embedding.shape)

# Display the first 10 values of the embedding
print("\nFirst 10 embedding values:")
print(query_embedding[:10])
```

```python
# Calculate cosine similarity between the query and all document embeddings
similarity_scores = cosine_similarity(
    query_embedding.reshape(1, -1),
    document_embeddings
)

# Display the similarity scores
print("Similarity scores:")
print(similarity_scores)

# Display the shape of the similarity scores
print("\nSimilarity scores shape:", similarity_scores.shape)
```

```python
# Sort document indices by similarity scores in descending order
ranked_indices = np.argsort(similarity_scores[0])[::-1]

# Select the top 3 document indices
top_3_indices = ranked_indices[:3]

# Display the top 3 matching documents
print("Top 3 Search Results:\n")

for rank, index in enumerate(top_3_indices, start=1):

    # Retrieve the corresponding document details
    document_id = df.iloc[index]["document_id"]
    document_text = df.iloc[index]["text"]
    score = similarity_scores[0][index]

    print(f"Rank: {rank}")
    print(f"Document ID: {document_id}")
    print(f"Text: {document_text}")
    print(f"Similarity Score: {score:.4f}")
    print("-" * 60)
```

```python
# Define a reusable semantic search function
def semantic_search(query, top_k=3):

    # Convert the user query into an embedding
    query_embedding = model.encode(
        query,
        convert_to_numpy=True
    )

    # Calculate similarity between query and all documents
    similarity_scores = cosine_similarity(
        query_embedding.reshape(1, -1),
        document_embeddings
    )[0]

    # Get document indices sorted by similarity (highest first)
    ranked_indices = np.argsort(similarity_scores)[::-1]

    # Select the top K matching documents
    top_indices = ranked_indices[:top_k]

    # Store the search results
    results = []

    # Collect details of each matching document
    for index in top_indices:

        results.append({
            "document_id": df.iloc[index]["document_id"],
            "text": df.iloc[index]["text"],
            "similarity_score": round(float(similarity_scores[index]), 4)
        })

    # Return the search results
    return pd.DataFrame(results)
```

```python
# Define a new search query
query = "How can I store information in a database?"

# Perform semantic search and retrieve top 3 results
results = semantic_search(query, top_k=3)

# Display the search results
display(results)
```

```python
# Define 10 test queries and their expected relevant document IDs
evaluation_queries = [
    {
        "query": "How do I write reusable code in Python?",
        "relevant_doc_ids": [4, 1]
    },
    {
        "query": "How can a Java program execute multiple tasks simultaneously?",
        "relevant_doc_ids": [10]
    },
    {
        "query": "Predict a continuous numerical value using data.",
        "relevant_doc_ids": [12]
    },
    {
        "query": "How can computers understand written human language?",
        "relevant_doc_ids": [17]
    },
    {
        "query": "Retrieve and manage records in relational databases.",
        "relevant_doc_ids": [21]
    },
    {
        "query": "Store flexible data in JSON-like document format.",
        "relevant_doc_ids": [22]
    },
    {
        "query": "Run applications using virtual machines in the cloud.",
        "relevant_doc_ids": [28]
    },
    {
        "query": "Store files and objects using AWS cloud storage.",
        "relevant_doc_ids": [27]
    },
    {
        "query": "Package applications with their required dependencies.",
        "relevant_doc_ids": [31, 32]
    },
    {
        "query": "Automatically build and test software after code changes.",
        "relevant_doc_ids": [35]
    }
]

# Display the total number of evaluation queries
print("Total evaluation queries:", len(evaluation_queries))
```

```python
# Store evaluation results
evaluation_results = []

# Process each evaluation query
for item in evaluation_queries:

    # Extract query and expected relevant document IDs
    query = item["query"]
    relevant_ids = item["relevant_doc_ids"]

    # Retrieve the top 3 search results
    results = semantic_search(query, top_k=3)

    # Extract retrieved document IDs
    retrieved_ids = results["document_id"].tolist()

    # Count relevant documents in the retrieved results
    relevant_count = sum(
        doc_id in relevant_ids for doc_id in retrieved_ids
    )

    # Calculate Precision@3
    precision = relevant_count / 3

    # Store evaluation details
    evaluation_results.append({
        "query": query,
        "retrieved_ids": retrieved_ids,
        "relevant_count": relevant_count,
        "precision_at_3": round(precision, 4)
    })

# Convert evaluation results into a DataFrame
evaluation_df = pd.DataFrame(evaluation_results)

# Display evaluation results
display(evaluation_df)
```

```python
# Calculate average Precision@3 across all queries
average_precision = evaluation_df["precision_at_3"].mean()

# Display the overall evaluation score
print("========== EVALUATION SUMMARY ==========")
print("Total Queries:", len(evaluation_df))
print("Average Precision@3:", round(average_precision, 4))
print("Precision@3 Percentage:", round(average_precision * 100, 2), "%")
```
